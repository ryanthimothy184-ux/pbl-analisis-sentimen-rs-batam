# **Scrapping berdasarkan search mencakup lebih dari 1 rs**

```python
from apify_client import ApifyClient
import pandas as pd

client = ApifyClient("apify_api_.........")  # Console > Settings > Integrations

rs_list = [
    "RS Awal Bros Batu Aji Batam",
    "RS Awal Bros Botania Batam",
    "RS Umum Awal Bros Batam Baloi",
    "RS Badan Pengusahaan Batam Sekupang",
    "RS Bhayangkara Batam Polda Kepri",
    "RS Bunda Asih Batam",
    "RS Griya Medika Batam",
    "RS Hj. Bunda Halimah Batam",
    "RS Ibu dan Anak Frisdhy Angel Batam",
    "RS Ibu dan Anak Kasih Sayang Ibu Batam",
    "RS Jasmine Batam",
    "RS Keluarga Husada Batam",
    "RS Mutiara Aini Batam",
    "RS Santa Elisabeth Sei Lekop Batam",
    "RS Umum Santa Elisabeth Batam Anggrek",
    "RS Umum Santa Elisabeth Batam Kota",
    "RS Umum Budi Kemuliaan Batam",
    "RS Umum Camatha Sahidya Batam",
    "RS Umum Charis Medika Batam",
    "RSUD Embung Fatimah Kota Batam",
    "RS Umum Graha Hermine Batam",
    "RS Umum Harapan Bunda Batam",
    "RS Umum Soedarsono Darmosoewito Batam",
    "RSIA Royal Medika Batam",
]

assert len(rs_list) == 24

run_input = {
    "searchStringsArray": rs_list,
    #"locationQuery": "Batam, Indonesia",
    "maxCrawledPlacesPerSearch": 10,
    "maxReviews": 3000,
    "reviewsSort": "newest",
    "scrapeReviewsPersonalData": True,
    "language": "id",
}

run = client.actor("compass/crawler-google-places").call(run_input=run_input)
items = list(client.dataset(run.default_dataset_id).iterate_items())

rows = []
for place in items:
    for r in place.get("reviews") or []:
        rows.append({
            "nama_tempat": place.get("title"),
            "bintang": r.get("stars"),
            "review": r.get("text"),
        })

df = pd.DataFrame(rows).dropna(subset=["review"]).drop_duplicates()
df.to_csv("reviews.csv", index=False, encoding="utf-8-sig")
print(df.shape)
```
